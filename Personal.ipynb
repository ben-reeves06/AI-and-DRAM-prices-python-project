```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf
import sys
```

```python
# importing datasets
GCIAI = pd.read_excel("AI_GCI.xlsx")
Shares = pd.read_excel("Shares.xlsx")
PCShip = pd.read_excel("PCShipment.xlsx")

```

```python
merged = pd.merge(GCIAI,Shares, on="Year")
```

```python
merged2 = pd.merge(merged, PCShip, on="Year")
```

```python
merged2.head()
```

```python
merged2 = merged2.rename(columns={"GCI in AI (billions)": "GCI_AI", "Shipments (Millions)": "Shipments"})
merged2.head()
```

```python
reg1 = smf.ols("DRAMPrice ~ Shipments + GCI_AI", data=merged2).fit()
reg2 = smf.ols("DRAMPrice ~ HBM_Share + DRAM_PerPC + Shipments", data=merged2).fit()
```

```python
print(reg1.summary())
```
